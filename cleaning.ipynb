```python
import pandas as pd
import numpy as np
```

```python
# 1. Load dataset
df = pd.read_csv("seattle-weather.csv")
```

```python
# 2. Convert date
df['date'] = pd.to_datetime(df['date'])

```

```python
# 3. Remove outliers (SAFE METHOD)
numeric_cols = df.select_dtypes(include=np.number).columns

mask = pd.Series(True, index=df.index)

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    mask &= (df[col] >= lower) & (df[col] <= upper)

df = df[mask]
```

```python
# 4. Reset index
df.reset_index(drop=True, inplace=True)
```

```python
# 5. Check result
print(df.shape)
print(df.head())
```

```python

# 6. Save cleaned file
df.to_csv("cleaned_seattle_weather.csv", index=False)
```
